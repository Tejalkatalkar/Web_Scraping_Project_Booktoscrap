# Step 1: Import Required Libraries

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
```

# Step 2: Define the URL of the website to scrape

```python
url = "https://books.toscrape.com/"
response = requests.get(url)
print("URL:", url)

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")
```

# Step 3: Checking Website Response

```python
if response.status_code == 200:
    # Status code 200 means the request was successful
    print("Website accessed successfully! Content retrieved.")
elif response.status_code == 404:
    # Status code 404 means the page was not found
    print("Page not found (404). Check the URL.")
else:
    # Other status codes indicate different issues
    # e.g., 403 = forbidden, 500 = server error
    print(f"Failed to access the website. Status code: {response.status_code}")

# print response status
print(response.status_code)
```

# Step 4: Save Extracted Data into CSV Format

```python
BASE_URL = "https://books.toscrape.com/catalogue/page-{}.html"
all_books = []

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
}

for page in range(1, 51):
    url = BASE_URL.format(page)
    try:
        response = requests.get(url, headers=headers, timeout=10)  # 10 sec timeout
        print(f"Page {page} status:", response.status_code)
    except requests.exceptions.RequestException as e:
        print(f"Error on page {page}:", e)
        continue  # skip this page

    if response.status_code != 200:
        break

    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.find_all("article", class_="product_pod")

    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        rating = book.p["class"][1]
        availability = book.find("p", class_="instock availability").text.strip()
        link = "https://books.toscrape.com/catalogue/" + book.h3.a["href"]
        all_books.append([title, price, rating, availability, link])
    
    time.sleep(1)  # delay 1 second between requests to avoid blocking

df = pd.DataFrame(all_books, columns=["Title", "Price", "Rating", "Availability", "Link"])
df.to_csv("books_data.csv", index=False, encoding="utf-8")
print("Scraping complete! Total rows:", len(df))
```

# Step 6 : Loop through each book

```python
# all <article> tags with class "product_pod"
books = soup.find_all("article", class_="product_pod")

for book in books:
    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text
    availability = book.find("p", class_="instock availability").text.strip()
    rating = book.p["class"][1]  # Get second class (rating)
    
    # Construct the full link to the book
    link = BASE_URL + book.h3.a["href"]

    print("Title:", title)
    print("Price:", price)
    print("Availability:", availability)
    print("Rating:", rating)
    print("Link:", link)
    print("-" * 50)  # Separator
```
